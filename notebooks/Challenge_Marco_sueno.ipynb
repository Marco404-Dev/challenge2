# Challenge: análisis y predicción del ciclo de sueño de bebés

**Caso:** estimar la duración de un ciclo de sueño (en minutos) a partir de la edad (en meses).

**Objetivo:** implementar y comparar dos soluciones en Python: **búsqueda del registro más cercano** y **predicción mediante regresión lineal**.

**Presentación:** ejecuta las celdas de arriba hacia abajo en Google Colab. Cambia la edad en la sección 4 para probar otro caso.

> **Importante:** los tres datos son aproximaciones tomadas de una gráfica del ejercicio, no mediciones clínicas. Este challenge es demostrativo y no permite inferir patrones reales de sueño ni realizar recomendaciones médicas.

## 1. Planteamiento del problema

Disponemos de tres pares de valores:

| Edad del bebé (meses) | Duración del ciclo (minutos) |
|---:|---:|
| 3 | 50 |
| 6 | 60 |
| 12 | 70 |

**Preguntas del challenge:** ¿cómo cambia la duración registrada con la edad? ¿qué dato es el más cercano a una edad nueva? ¿qué valor estima una recta ajustada?

```python
import numpy as np
import matplotlib.pyplot as plt

edades = np.array([3, 6, 12], dtype=float)
ciclos = np.array([50, 60, 70], dtype=float)

print("DATOS DEL CASO")
for edad, ciclo in zip(edades, ciclos):
    print(f"{edad:g} meses -> {ciclo:g} minutos")
```

## 2. Razonamiento: identificar la tendencia

Calculamos la diferencia entre los registros consecutivos. **Atención:** los intervalos de edad son distintos (3 y 6 meses); no se debe confundir un incremento absoluto con una tasa mensual.

```python
print("ANÁLISIS DE LOS REGISTROS")
for i in range(1, len(edades)):
    diferencia_ciclo = ciclos[i] - ciclos[i-1]
    diferencia_edad = edades[i] - edades[i-1]
    tasa = diferencia_ciclo / diferencia_edad
    print(f"De {edades[i-1]:g} a {edades[i]:g} meses: +{diferencia_ciclo:g} min "
          f"({tasa:.2f} min/mes)")
print("Conclusión: en los tres registros del ejemplo, la duración aumenta con la edad.")
```

## 3. Dos enfoques de solución

**A. Búsqueda del vecino más cercano:** identifica la edad registrada con menor distancia a la edad consultada y devuelve su valor original.

**B. Regresión lineal:** ajusta una recta `duración = pendiente × edad + intercepto` y calcula un valor estimado para la edad consultada. No reproduce necesariamente cada dato observado.

```python
def buscar_registro_cercano(edad_consulta):
    indice = int(np.argmin(np.abs(edades - edad_consulta)))
    return edades[indice], ciclos[indice]

pendiente, intercepto = np.polyfit(edades, ciclos, 1)

def predecir_ciclo(edad_consulta):
    return pendiente * edad_consulta + intercepto

print(f"Modelo lineal: duración = {pendiente:.4f} × edad + {intercepto:.4f}")
```

## 4. Ejecutar el challenge

Cambia **`edad_consulta`** y ejecuta desde esta celda hasta el final. El intervalo **3 a 12 meses** corresponde al rango de los datos usados para el ajuste.

```python
# CAMBIA ESTE VALOR PARA TU DEMOSTRACIÓN
edad_consulta = 8  # meses, entre 3 y 12

if not (3 <= edad_consulta <= 12):
    raise ValueError("La edad debe estar entre 3 y 12 meses para este challenge.")

edad_cercana, ciclo_cercano = buscar_registro_cercano(edad_consulta)
ciclo_estimado = predecir_ciclo(edad_consulta)

print(f"Edad consultada: {edad_consulta:g} meses")
print(f"Búsqueda: registro de {edad_cercana:g} meses -> {ciclo_cercano:g} min")
print(f"Predicción lineal: {ciclo_estimado:.2f} min")
print(f"Diferencia entre enfoques: {abs(ciclo_cercano-ciclo_estimado):.2f} min")
```

## 5. Visualizar y comparar

Los puntos azules representan los valores originales; el punto naranja es el registro más cercano y el rojo es la predicción obtenida con la recta.

```python
x = np.linspace(edades.min(), edades.max(), 200)
plt.figure(figsize=(9, 5))
plt.scatter(edades, ciclos, s=85, color="royalblue", label="Registros aproximados", zorder=4)
plt.plot(x, predecir_ciclo(x), color="seagreen", linewidth=2,
         label="Ajuste lineal")
plt.scatter([edad_cercana], [ciclo_cercano], color="darkorange", s=180,
            marker="X", label="Registro más cercano", zorder=5)
plt.scatter([edad_consulta], [ciclo_estimado], color="crimson", s=125,
            label="Predicción consultada", zorder=6)
plt.axvline(edad_consulta, linestyle="--", color="gray", alpha=0.55)
plt.title("Challenge: búsqueda frente a predicción del ciclo de sueño")
plt.xlabel("Edad (meses)")
plt.ylabel("Duración del ciclo (minutos)")
plt.xticks(sorted(set([3, 6, 12, edad_consulta])))
plt.grid(alpha=0.2)
plt.legend()
plt.tight_layout()
plt.show()
```

## 6. Comprobación rápida

Validamos que la búsqueda devuelve un registro existente, que la regresión produce un número y que los límites elegidos están cubiertos.

```python
assert buscar_registro_cercano(3) == (3.0, 50.0)
assert buscar_registro_cercano(12) == (12.0, 70.0)
assert np.isfinite(predecir_ciclo(8))
assert edades.min() <= edad_consulta <= edades.max()
print("Pruebas básicas superadas.")
```

## 7. Resultados y conclusión

- **Búsqueda:** entrega un valor observado del conjunto, asociado a la edad más cercana.
- **Predicción:** entrega una estimación continua con una recta ajustada a los tres puntos.
- **Diferencia:** el vecino más cercano utiliza una observación; la recta incorpora la tendencia general del conjunto.
- **Limitación:** solo hay tres datos aproximados y no se ha validado el modelo con datos independientes. La gráfica demuestra el procedimiento, **no** una predicción confiable para bebés reales.
